### 3.2 Line Pots (Series/Dataframe)

What is a line plot and why use it?

A line chart or line plot is a type of plot which displays information as a series of data points called 'markers' connected by straight line segments. It is a basic type of chart common in many fields. Use line plot when you have a continuous data set. These are best suited for trend-based visualizations of data over a period of time.

**Let's start with a case study:**

In 2010, Haiti suffered a catastrophic magnitude 7.0 earthquake. The quake caused widespread devastation and loss of life and about three million people were affected by this natural disaster. As part of Canada's humanitarian effort, the Government of Canada stepped up its effort in accepting refugees from Haiti. We can quickly visualize this effort using a `Line` plot:

**Question:** Plot a line graph of immigration from Haiti using `df.plot()`.


```python
#Since we converted the years to string,
#let's declare a variable that will allow us to easily call upon the full range of years:
years = list(map(str, range(1980, 2014)))
#creating data series
haiti = df_can.loc['Haiti', years] # passing in years 1980 - 2013 to exclude the 'total' column
haiti.head()
```

```python
haiti.plot()
```

*pandas* automatically populated the x-axis with the index values (years), and the y-axis with the column values (population). <br>

Also, let's label the x and y axis using `plt.title()`, `plt.ylabel()`, and `plt.xlabel()` as follows:


```python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show() # need this line to show the updates made to the figure
```

### 3.3 Adding Annotation to plot

We can clearly notice how number of immigrants from Haiti spiked up from 2010 as Canada stepped up its efforts to accept refugees from Haiti. Let's annotate this spike in the plot by using the `plt.text()` method.<br>
<br>
However, notice that years are of type *string*. Let's change the type of the index values to *integer* first.


```python
haiti.index = haiti.index.map(int)
#Converting to int. I can use them directly on the x-axis and for annotations. If years were stored as strings, Matplotlib wouldn't know how to interpret '2000' as a number. x=2000: Horizontal position (left/right placement) and y=6000: Vertical position (up/down placement).
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# annotate the 2010 Earthquake.
# syntax: plt.text(x, y, label)
plt.text(2000, 6000, '2010 Earthquake') # see note below

plt.show()
```

```python
#Restore Year Labels on the X-Axis
haiti.index = haiti.index.map(int)  # Convert index back to integers
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# Fix the x-axis labels
plt.ticklabel_format(style='plain', axis='x')  # Disable scientific notation
plt.xticks(ticks=haiti.index, labels=haiti.index, rotation=45)  # Use year values as x-axis labels

# Annotate the 2010 Earthquake
plt.text(2000, 6000, '2010 Earthquake')  # Place the annotation

plt.show()
```

### 3.4 Improving readability

```python
plt.figure(figsize=(12, 6))  # Increase the figure size

haiti.index = haiti.index.map(int)  # Convert index back to integers
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# Fix the x-axis labels and prevent scientific notation
plt.ticklabel_format(useOffset=False, style='plain', axis='x')  # Disable scientific notation
plt.xticks(ticks=range(1980, 2014, 5))  # Show every 5th year

# Annotate the 2010 Earthquake
plt.text(2004, 6050, '2010 Earthquake')  # Place the annotation
plt.scatter(2010, 6000, color='red', zorder=5)  # Add a red dot at (2010, 6000)

plt.show()
```

## 4.Question: Let's compare the number of immigrants from India and China from 1980 to 2013.

```python
df_CI = df_can.loc[['India', 'China'], years]
df_CI
```

```python
df_CI.plot(kind='line') #Plot graph. We will explicitly specify line plot by passing in kind parameter to plot()
```

### 4.1 Improving readability

```python
# Transpose the DataFrame so years are on the x-axis
df_CI = df_CI.T  # Switch rows and columns

# Convert the index (years) to integers for proper plotting
df_CI.index = df_CI.index.map(int)

# Plot the data for India and China
df_CI.plot(kind='line', figsize=(12, 6))

# Add titles and labels
plt.title('Immigration from China and India')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# Fix x-axis to show proper years
plt.ticklabel_format(useOffset=False, style='plain', axis='x')  # Disable scientific notation
plt.xticks(ticks=range(1980, 2014, 5))  # Show every 5th year for better readability

# Show the plot
plt.show()
```
