Let's download and import our primary Canadian Immigration dataset using pandas's read_excel() method.

````python
df_can = pd.read_excel(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.xlsx',
    sheet_name='Canada by Citizenship',
    skiprows=range(20),
    skipfooter=2)

print('Data read into a pandas dataframe!')
````

###2.3 Clean Data

Let's clean the data set to remove a few unnecessary columns. We can use pandas drop() method as follows:

````python
# in pandas axis=0 represents rows (default) and axis=1 represents columns.
df_can.drop(['AREA','REG','DEV','Type','Coverage'], axis=1, inplace=True)
df_can.head(5)
````

###2.4 Reanme Colums

Let's rename the columns so that they make sense. We can use rename() method by passing in a dictionary of old and new names as follows:

````python
df_can.rename(columns={'OdName':'Country', 'AreaName':'Continent', 'RegName':'Region'}, inplace=True)
df_can.columns
````

###2.4 Add Total column

We will also add a 'Total' column that sums up the total immigrants by country over the entire period 1980 - 2013, as follows:

````python
df_can['Total'] = df_can.sum(axis=1)
df_can['Total']
````

###2.5 Solving TypeError

Lets solve this small issue

````python
print(df_can.dtypes)
````

Exclude Non-Numeric Columns

````python
#Exclude Non-Numeric Columns

df_can['Total'] = df_can.select_dtypes(include='number').sum(axis=1)
df_can['Total']
````

We can check to see how many null objects we have in the dataset as follows:

````python
#We can check to see how many null objects we have in the dataset as follows:
df_can.isnull().sum()
````

##  3.Indexing and Selection (slicing)

**There are two ways to filter on a column name:**

Method 1: Quick and easy, but only works if the column name does NOT have spaces or special characters.
```python
    df.column_name               # returns series
```

Method 2: More robust, and can filter on multiple columns.

```python
    df['column']                  # returns series
```

```python
    df[['column 1', 'column 2']]  # returns dataframe
```
---


````python
#Let's try filtering on the list of countries ('Country')
df_can.Country  # returns a series
````

### 3.2 Select Rows

There are main 2 ways to select rows:

```python
    df.loc[label]    # filters by the labels of the index/column
    df.iloc[index]   # filters by the positions of the index/column
```


````python
df_can.head(1)
````

````python
#Setting index to Column "Country" / To reset index we can use df_can.reset_index()
df_can.set_index('Country', inplace=True)
````

````python
df_can.head(1)
````

````python
# optional: to remove the name of the index. I can always add it back: df_can.index.name = 'Country'
df_can.index.name = None
````

````python
df_can.head(1)
````

````python
# 1. the full row data (all columns)
df_can.loc['Japan']
````

````python
# alternate methods
df_can.iloc[87]
````

````python
df_can[df_can.index == 'Japan']
````

````python
# 2. for year 2013
df_can.loc['Japan', 2013]
````

````python
# alternate method
# year 2013 is the last column, with a positional index of 36
df_can.iloc[87, 36]
````

````python
# 3. for years 1980 to 1985
df_can.loc['Japan', [1980, 1981, 1982, 1983, 1984, 1984]]
````
