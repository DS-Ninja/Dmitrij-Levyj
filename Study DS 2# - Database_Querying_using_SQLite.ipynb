## Problem 2: List community area names and numbers with per capita income less than 11000.

```python
# Connecting to SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database file if different

# SQL query to find community area names and numbers with per capita income < 11000
query = """
SELECT community_area_number, community_area_name
FROM CENSUS_DATA
WHERE per_capita_income < 11000;
"""

# Executing the query and load the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Display the results
print(results)

# Closing the database connection
conn.close()
```

## Problem 4: List all kidnapping crimes involving a child?

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find all kidnapping crimes involving a child
query = """
SELECT case_number, description, date, block
FROM CHICAGO_CRIME_DATA
WHERE primary_type = 'KIDNAPPING'
  AND description LIKE '%Child%';
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the results
print(results)

# Closing the connection to the SQLite database
conn.close()
```

## Problem 5: List the kind of crimes that were recorded at schools. (No repetitions)

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find distinct types of crimes recorded at schools
query = """
SELECT DISTINCT primary_type
FROM CHICAGO_CRIME_DATA
WHERE location_description LIKE '%School%';
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the results
print("Unique kinds of crimes recorded at schools:")
print(results)

# Closing the connection to the SQLite database
conn.close()
```

## Problem 6: List the type of schools along with the average safety score for each type.

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find the average safety score for each type of school
query = """
SELECT "Elementary, Middle, or High School" AS school_type,
       AVG(safety_score) AS average_safety_score
FROM CHICAGO_PUBLIC_SCHOOLS
GROUP BY "Elementary, Middle, or High School";
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the results
print("School types along with their average safety scores:")
print(results)

# Closing the connection to the SQLite database
conn.close()
```

## Problem 7: List 5 community areas with highest % of households below poverty line

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find 5 community areas with the highest % of households below poverty line
query = """
SELECT COMMUNITY_AREA_NAME, COMMUNITY_AREA_NUMBER, percent_households_below_poverty AS poverty_rate
FROM CENSUS_DATA
ORDER BY percent_households_below_poverty DESC
LIMIT 5;
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the results
print("Top 5 community areas with highest % of households below poverty line:")
print(results)

# Closing the connection to the SQLite database
conn.close()
```

## Problem 8: Which community area is most crime prone? Display the coumminty area number only.

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find the community area number with the most crimes
query = """
SELECT COMMUNITY_AREA_NUMBER
FROM CHICAGO_CRIME_DATA
GROUP BY COMMUNITY_AREA_NUMBER
ORDER BY COUNT(*) DESC
LIMIT 1;
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the most crime-prone community area number
print("Community area number most prone to crime:")
print(results)
```

## Problem 9: Use a sub-query to find the name of the community area with highest hardship index

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find the name of the community area with the highest hardship index using a subquery
query = """
SELECT COMMUNITY_AREA_NAME
FROM CENSUS_DATA
WHERE hardship_index = (SELECT MAX(hardship_index) FROM CENSUS_DATA);
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the community area name with the highest hardship index
print("Community area with the highest hardship index:")
print(results)

# Closing the connection to the SQLite database
conn.close()
```

## Problem 10: Use a sub-query to determine the Community Area Name with most number of crimes?

```python
# Connecting to the SQLite database
conn = sqlite3.connect('FinalDB.db')  # Replace 'FinalDB.db' with your database name

# SQL query to find the community area name with the most number of crimes using a subquery
query = """
SELECT COMMUNITY_AREA_NAME
FROM CENSUS_DATA
WHERE COMMUNITY_AREA_NUMBER = (
    SELECT COMMUNITY_AREA_NUMBER
    FROM CHICAGO_CRIME_DATA
    GROUP BY COMMUNITY_AREA_NUMBER
    ORDER BY COUNT(*) DESC
    LIMIT 1
);
"""

# Executing the query and loading the results into a pandas DataFrame
results = pd.read_sql(query, conn)

# Displaying the community area name with the most number of crimes
print("Community area with the most number of crimes:")
print(results)

# Closing the connection to the SQLite database
conn.close()
```
