#2.2 Data Scraping

````python
# Initialize variables
base_url = "https://www.cvbankas.lt/?padalinys[0]=76&page="
job_links = []

# Iterate through pages 1 to 6
for page in range(1, 7):
    url = f"{base_url}{page}"
    response = requests.get(url)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        # Find all job links
        links = soup.find_all("a", class_="list_a can_visited list_a_has_logo")
        for link in links:
            job_links.append(link.get('href'))

# Create a DataFrame for the job links
df_jobs = pd.DataFrame(job_links, columns=["Job Link"])

# Save to a CSV file
df_jobs.to_csv("job_links.csv", index=False)

print("Job links saved to 'job_links.csv'")
````

````python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Load the job links from the uploaded CSV
file_path = '/content/job_links.csv'  # Update the file path if necessary
job_links_df = pd.read_csv(file_path)

# Initialize a list to store job details
job_details = []

# Iterate over each job link
for job_link in job_links_df['Job Link']:
    response = requests.get(job_link)
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        # Extract job position
        position = soup.find("h1", class_="heading1", id="jobad_heading1")
        position = position.text.strip() if position else None

        # Extract salary
        salary = soup.find("span", class_="salary_amount")
        salary = salary.text.strip() if salary else None

        # Extract job description
        description_section = soup.find("section", itemprop="description")
        description = description_section.text.strip() if description_section else None

        # Append details to the list
        job_details.append({
            "Position": position,
            "Salary (EUR)": salary,
            "Description": description
        })

# Convert the list to a DataFrame
job_details_df = pd.DataFrame(job_details)

# Save to a CSV file
job_details_df.to_csv("job_details.csv", index=False)

print("Job details saved to 'job_details.csv'")
````
