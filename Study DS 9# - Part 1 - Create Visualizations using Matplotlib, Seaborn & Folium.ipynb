### 2.2 Data Visualizations


#### TASK 1.1: Develop a *Line chart* using the functionality of pandas to show how automobile sales fluctuate from year to year


````python
# Group by Year and calculate mean sales
df_line = df.groupby('Year')['Automobile_Sales'].mean()

# Create the line plot
plt.figure(figsize=(10, 6))  # Set figure size
df_line.plot(kind='line')    # Create a line plot
plt.xlabel('Year')           # Label for x-axis
plt.ylabel('Average Automobile Sales')  # Label for y-axis
plt.title('Yearly Fluctuation in Automobile Sales')  # Title
plt.grid(True)  # Optional: Add grid for better readability
plt.show()
````

````python
# Group by Year and calculate mean sales (if applicable; not needed here for a single dataset)
df_line = df.groupby('Year')['Automobile_Sales'].mean()

# Create the plot
plt.figure(figsize=(10, 6))
df_line.plot(kind='line')

# Add x-ticks and labels
plt.xticks(list(range(1980, 1985)), rotation=75)  # Adjust x-ticks for your data range
plt.xlabel('Year')  # Label for the x-axis
plt.ylabel('Automobile Sales')  # Label for the y-axis
plt.title('Yearly Automobile Sales Trends')  # Title of the chart

# Add annotations
plt.text(1982, 650, '1981-82 Recession')  # Annotation for 1981-82 recession
plt.text(1984, 800, 'Economic Recovery', fontsize=10)  # Example annotation

# Add legend (optional; specify a label in the plot call if required)
plt.legend(['Average Sales'], loc='upper left')  # Adjust legend label and position as needed

# Show the plot
plt.show()
````

####TASK 1.2: Plot different lines for categories of vehicle type and analyse the trend to answer the question Is there a noticeable difference in sales trends between different vehicle types during recession periods?¶

````python
# Step 1: Filter for recession years
df_rec = df[df['Recession'] == 1]

# Step 2: Group by Year and Vehicle_Type, calculate average sales
df_Mline = df_rec.groupby(['Year', 'Vehicle_Type'], as_index=False)['Automobile_Sales'].mean()

# Step 3: Normalize sales for each vehicle type
df_Mline['Normalized_Sales'] = df_Mline.groupby('Vehicle_Type')['Automobile_Sales'].transform(lambda x: x / x.mean())

# Step 4: Set Year as index for plotting
df_Mline.set_index('Year', inplace=True)

# Step 5: Create the plot for each vehicle type
plt.figure(figsize=(12, 8))
for vehicle_type in df_Mline['Vehicle_Type'].unique():
    data = df_Mline[df_Mline['Vehicle_Type'] == vehicle_type]
    plt.plot(data.index, data['Normalized_Sales'], label=vehicle_type, marker='o')

# Step 6: Highlight recession years
recession_years = df_rec['Year'].unique()
for year in recession_years:
    plt.axvline(x=year, color='gray', linestyle='--', alpha=0.5)

# Step 7: Add labels, legend, and title
plt.legend(title="Vehicle Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.ylabel("Normalized Sales")
plt.xlabel("Year")
plt.title("Normalized Automobile Sales by Vehicle Type During Recession")

# Finalize the plot
plt.tight_layout()
plt.show()
````

####TASK 1.3: Use the functionality of Seaborn Library to create a visualization to compare the sales trend per vehicle type for a recession period with a non-recession period.

````python
# Step 1: Group by Recession and Vehicle_Type to calculate average sales
new_df = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

# Step 2: Create the bar chart using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Recession', y='Automobile_Sales', hue='Vehicle_Type', data=new_df, palette="muted")

# Step 3: Add labels and title
plt.xlabel('Recession (0 = Non-Recession, 1 = Recession)')
plt.ylabel('Average Automobile Sales')
plt.title('Average Automobile Sales During Recession and Non-Recession Periods')
plt.xticks(ticks=[0, 1], labels=['Non-Recession', 'Recession'])

# Step 4: Show the plot
plt.tight_layout()
plt.show()
````

````python
# Step 1: Group by Recession and Vehicle_Type, calculate average sales
dd = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()

# Step 2: Create the grouped bar chart using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Recession', y='Automobile_Sales', hue='Vehicle_Type', data=dd, palette="muted")

# Step 3: Customize the chart
plt.xticks(ticks=[0, 1], labels=['Non-Recession', 'Recession'])
plt.xlabel('Recession (0 = Non-Recession, 1 = Recession)')
plt.ylabel('Average Automobile Sales')
plt.title('Vehicle-Wise Sales During Recession and Non-Recession Periods')

# Step 4: Show the plot
plt.tight_layout()
plt.show()
````

####TASK 1.4: Use sub plotting to compare the variations in GDP during recession and non-recession period by developing line plots for each period.


````python
# Create dataframes for recession and non-recession periods
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]

# Create the figure and subplots
fig = plt.figure(figsize=(12, 6))

# Subplot 1: GDP Variation during Recession Period
ax0 = fig.add_subplot(1, 2, 1)  # 1 row, 2 columns, first plot
sns.lineplot(x='Year', y='GDP', data=rec_data, label='Recession', ax=ax0)
ax0.set_xlabel('Year')
ax0.set_ylabel('GDP')
ax0.set_title('GDP Variation During Recession Period')
ax0.grid(visible=True, linestyle='--', alpha=0.7)

# Subplot 2: GDP Variation during Non-Recession Period
ax1 = fig.add_subplot(1, 2, 2)  # 1 row, 2 columns, second plot
sns.lineplot(x='Year', y='GDP', data=non_rec_data, label='Non-Recession', ax=ax1)
ax1.set_xlabel('Year')
ax1.set_ylabel('GDP')
ax1.set_title('GDP Variation During Non-Recession Period')
ax1.grid(visible=True, linestyle='--', alpha=0.7)

# Adjust layout for better spacing
plt.tight_layout()

# Display the plots
plt.show()
````

####TASK 1.5: Develop a Bubble plot for displaying the impact of seasonality on Automobile Sales.


````python
# Filter non-recession data
non_rec_data = df[df['Recession'] == 0]

# Bubble plot using Seaborn's scatterplot
plt.figure(figsize=(12, 8))
sns.scatterplot(
    data=non_rec_data,
    x='Month',
    y='Automobile_Sales',
    size='Seasonality_Weight',  # Bubble size
    hue='Seasonality_Weight',  # Bubble color
    legend='brief',
    sizes=(50, 500)  # Adjust bubble size range
)

# Customize the plot
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.title('Seasonality Impact on Automobile Sales')

# Show the plot
plt.tight_layout()
plt.show()
````

####TASK 1.6: Use the functionality of Matplotlib to develop a scatter plot to identify the correlation between average vehicle price relate to the sales volume during recessions.


````python
# Step 1: Create a dataframe for recession periods
rec_data = df[df['Recession'] == 1]

# Step 2: Create a scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(rec_data['Consumer_Confidence'], rec_data['Automobile_Sales'], color='blue', alpha=0.7)

# Step 3: Customize the plot
plt.xlabel('Consumer Confidence')
plt.ylabel('Automobile Sales')
plt.title('Consumer Confidence and Automobile Sales during Recessions')

# Step 4: Show the plot
plt.grid(visible=True, linestyle='--', alpha=0.5)
plt.show()
````

####TASK 1.7: Create a pie chart to display the portion of advertising expenditure of XYZAutomotives during recession and non-recession periods.

````python
# Step 1: Filter the data for recession and non-recession periods
Rdata = df[df['Recession'] == 1]
NRdata = df[df['Recession'] == 0]

# Step 2: Calculate the total advertising expenditure for both periods
RAtotal = Rdata['Advertising_Expenditure'].sum()
NRAtotal = NRdata['Advertising_Expenditure'].sum()

# Step 3: Create a pie chart
plt.figure(figsize=(8, 6))

labels = ['Recession', 'Non-Recession']
sizes = [RAtotal, NRAtotal]
colors = ['#FF9999', '#66B3FF']  # Custom colors for the pie chart
explode = (0.1, 0)  # Slightly explode the "Recession" slice for emphasis

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',  # Display percentage values
    startangle=90,      # Start the pie chart at 90 degrees
    colors=colors,      # Use custom colors
    explode=explode     # Explode one slice
)

# Add title to the chart
plt.title('Advertising Expenditure during Recession and Non-Recession Periods')

# Display the chart
plt.show()
````

####TASK 1.8: Develop a pie chart to display the total Advertisement expenditure for each vehicle type during recession period.

````python
# Step 1: Filter the data for recession periods
Rdata = df[df['Recession'] == 1]

# Step 2: Calculate the total advertising expenditure by vehicle type during recessions
VTexpenditure = Rdata.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()

# Step 3: Create a pie chart
plt.figure(figsize=(8, 6))

labels = VTexpenditure.index
sizes = VTexpenditure.values
colors = ['#FF9999', '#66B3FF', '#99FF99']  # Custom colors for each slice

# Match the length of 'explode' with the number of slices
explode = [0.1] * len(sizes)  # Explode all slices slightly for emphasis

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%',  # Display percentage values
    startangle=90,      # Start the chart at 90 degrees
    colors=colors,      # Add custom colors
    explode=explode     # Explode all slices
)

# Add a title
plt.title('Share of Each Vehicle Type in Total Expenditure during Recessions')

# Display the chart
plt.show()
````

####TASK 1.10 Create a map on the hightest sales region/offices of the company during recession period

````python
from folium.plugins import MarkerCluster
````

````python
# Step 1: Filter the data for the recession period
recession_data = df[df['Recession'] == 1]

# Step 2: Calculate the total sales by "City" (actually state in this case)
sales_by_state = recession_data.groupby('City', as_index=False)['Automobile_Sales'].sum()

# Step 3: Add latitude and longitude for states
state_coords = {
    'California': [36.7783, -119.4179],
    'Georgia': [32.1656, -82.9001],
    'Illinois': [40.6331, -89.3985],
    'New York': [43.2994, -74.2179]
}

# Assign coordinates dynamically
sales_by_state['Latitude'] = sales_by_state['City'].map(lambda state: state_coords[state][0])
sales_by_state['Longitude'] = sales_by_state['City'].map(lambda state: state_coords[state][1])

# Step 4: Create a base map centered on the United States
map1 = folium.Map(location=[37.0902, -95.7129], zoom_start=4)

# Step 5: Add markers for each state with sales information
marker_cluster = MarkerCluster().add_to(map1)

for _, row in sales_by_state.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=f"{row['City']}: {row['Automobile_Sales']} units",
        icon=folium.Icon(color="blue", icon="info-sign")
    ).add_to(marker_cluster)

# Display the map
map1
````
